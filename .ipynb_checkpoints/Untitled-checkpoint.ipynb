```python
def load_single_housing_file_clean(file_path):
    """Load a single housing file with proper cleaning"""
    
    # Get filename info
    filename = os.path.basename(file_path)
    year = int(filename.split('_')[0])
    month = int(filename.split('_')[1])
    
  
    df = pd.read_csv(file_path, encoding='latin-1', skiprows=2, index_col=0)
    
 
    df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
    
   
    df = df.reset_index()
    df.columns = ['Zone', 'Single', 'Semi_Detached', 'Row', 'Apartment', 'All']
    

    metadata_keywords = ['Calgary', 'Notes', 'Source', 'CMA,', 'definitions', 'based on']
    
    def is_metadata_row(zone_name):
        if pd.isna(zone_name):
            return True
        zone_str = str(zone_name).strip()
        return any(keyword in zone_str for keyword in metadata_keywords)
    
    df = df[~df['Zone'].apply(is_metadata_row)].copy()
    
    # Convert numeric columns
    for col in ['Single', 'Semi_Detached', 'Row', 'Apartment', 'All']:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    df = df.fillna(0)
    
    # Only keep rows where zone name looks like actual zone name
    df = df[df['Zone'].str.len() > 2].copy()  
    df = df[df['All'] >= 0].copy() 
    
    # Add date info
    df['Year'] = year
    df['Month'] = month
    df['YearMonth'] = f"{year}-{month:02d}"
    
    return df


df_jan_2023_clean = load_single_housing_file_clean(file_path)

print("CLEANED JANUARY 2023 DATA:")
print(df_jan_2023_clean[['Zone', 'Single', 'Apartment', 'All']].to_string(index=False))

print(f"CLEAN INSIGHTS:")
print(f"Number of zones: {len(df_jan_2023_clean)}")
print(f"Total housing starts: {df_jan_2023_clean['All'].sum():.0f}")
print(f"Zones included: {list(df_jan_2023_clean['Zone'])}")
```

```python
# Load all housing starts files
housing_folder = r"C:\Users\Camille\Documents\GitHub\Camilleelot.github.io\Housing_Data_Project_CMHC\housing_starts"
csv_files = glob.glob(os.path.join(housing_folder, "*.csv"))

print(f"📁 Found {len(csv_files)} housing data files")

# Load all data
all_housing_data = []
for file in csv_files:
    df = load_single_housing_file_clean(file)
    if df is not None and len(df) > 0:
        all_housing_data.append(df)

# Combine all data
combined_housing = pd.concat(all_housing_data, ignore_index=True)
combined_housing = combined_housing.sort_values(['Year', 'Month', 'Zone'])

print(f"🎯 COMPLETE DATASET LOADED:")
print(f"Total records: {len(combined_housing)}")
print(f"Date range: {combined_housing['YearMonth'].min()} to {combined_housing['YearMonth'].max()}")
print(f"Unique zones: {len(combined_housing['Zone'].unique())}")
```

```python
# Create monthly summary
monthly_totals = combined_housing.groupby('YearMonth').agg({
    'Single': 'sum',
    'Semi_Detached': 'sum',
    'Row': 'sum', 
    'Apartment': 'sum',
    'All': 'sum'
}).reset_index()

# Sort chronologically
monthly_totals = monthly_totals.sort_values('YearMonth')

# Define the rezoning date (August 2024)
rezoning_date = '2024-08'

# Split before/after
before_rezoning = monthly_totals[monthly_totals['YearMonth'] < rezoning_date]
after_rezoning = monthly_totals[monthly_totals['YearMonth'] >= rezoning_date]

print("🏗️ CALGARY REZONING POLICY IMPACT ANALYSIS:")
print(f"\n📊 BEFORE REZONING (pre-August 2024):")
print(f"  Average monthly total starts: {before_rezoning['All'].mean():.0f}")
print(f"  Average monthly apartments: {before_rezoning['Apartment'].mean():.0f}")
print(f"  Average monthly single homes: {before_rezoning['Single'].mean():.0f}")

print(f"\n🏠 AFTER REZONING (August 2024+):")
print(f"  Average monthly total starts: {after_rezoning['All'].mean():.0f}")
print(f"  Average monthly apartments: {after_rezoning['Apartment'].mean():.0f}")
print(f"  Average monthly single homes: {after_rezoning['Single'].mean():.0f}")

# Calculate percentage changes
total_change = ((after_rezoning['All'].mean() - before_rezoning['All'].mean()) / before_rezoning['All'].mean()) * 100
apt_change = ((after_rezoning['Apartment'].mean() - before_rezoning['Apartment'].mean()) / before_rezoning['Apartment'].mean()) * 100

print(f"\n🎯 POLICY IMPACT:")
print(f"  Total housing starts changed by: {total_change:.1f}%")
print(f"  Apartment construction changed by: {apt_change:.1f}%")

print(f"\n📅 MONTHLY BREAKDOWN:")
print(monthly_totals[['YearMonth', 'Single', 'Apartment', 'All']].to_string(index=False))
```

```python
from scipy import stats
import numpy as np

# Create the clean enhanced density chart
fig, ax = plt.subplots(figsize=(14, 8))

# Prepare data
monthly_totals['Dense_Housing'] = monthly_totals['Apartment'] + monthly_totals['Row']
monthly_totals['Density_Ratio'] = monthly_totals['Dense_Housing'] / monthly_totals['Single']

# Create numeric x-axis for correlation
monthly_totals['Date_Numeric'] = range(len(monthly_totals))

# Split data
before_data = monthly_totals[monthly_totals['YearMonth'] < '2024-08'].copy()
after_data = monthly_totals[monthly_totals['YearMonth'] >= '2024-08'].copy()

# Plot the main data points
ax.plot(monthly_totals['YearMonth'], monthly_totals['Density_Ratio'], 
        marker='D', linewidth=3, markersize=6, color='purple', label='Density Ratio')

# Calculate and plot BEFORE trendline
if len(before_data) > 1:
    before_slope, before_intercept, before_r, _, _ = stats.linregress(
        before_data['Date_Numeric'], before_data['Density_Ratio'])
    
    before_line = before_slope * before_data['Date_Numeric'] + before_intercept
    ax.plot(before_data['YearMonth'], before_line, '--', linewidth=3, color='red', alpha=0.8,
            label=f'Before Trend (slope: {before_slope:.3f})')

# Calculate and plot AFTER trendline  
if len(after_data) > 1:
    after_slope, after_intercept, after_r, _, _ = stats.linregress(
        after_data['Date_Numeric'], after_data['Density_Ratio'])
    
    after_line = after_slope * after_data['Date_Numeric'] + after_intercept
    ax.plot(after_data['YearMonth'], after_line, '--', linewidth=3, color='lightgreen', alpha=0.8,
            label=f'After Trend (slope: {after_slope:.3f})')

# Add the rezoning line
ax.axvline(x='2024-08', color='red', linestyle='-', linewidth=3, alpha=0.9, 
           label='Rezoning Policy')

# Clean styling
ax.set_title('Density Revolution: Multi-Family vs Single Family\nwith Trend Analysis', 
             fontweight='bold', fontsize=16, pad=20)
ax.set_ylabel('Density Ratio (Dense Housing/Single Family)', fontsize=12)
ax.set_xlabel('Time Period', fontsize=12)
ax.grid(True, alpha=0.3)
ax.tick_params(axis='x', rotation=45)
ax.legend(loc='upper left', fontsize=11)

plt.tight_layout()
plt.show()

# Print stats separately
print("📊 TREND ANALYSIS:")
print(f"Before Rezoning: Slope {before_slope:.4f}, R-value {before_r:.3f}")
print(f"After Rezoning: Slope {after_slope:.4f}, R-value {after_r:.3f}")
print(f"🎯 Complete trend reversal: {((after_slope - before_slope) / abs(before_slope) * 100):.0f}% change in slope")
```

```python
def load_rental_data_fixed(file_path):
    """Load rental data with correct 7-column structure"""
    
    # Extract year from filename
    filename = os.path.basename(file_path)
    year = 2000 + int(filename.split()[1])
    
    try:
        # Load with proper structure
        df = pd.read_csv(file_path, encoding='latin-1', skiprows=2)
        df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
        
        # Remove metadata rows
        df = df[~df.iloc[:, 0].astype(str).str.contains('Notes|CMA|Source|definitions', na=False)]
        df = df[df.iloc[:, 0] != 'Calgary'].dropna(how='all')
        
        # Set correct column names for 7 columns
        df.columns = ['Zone', 'Vacancy_Rate', 'Availability_Rate', 'Avg_Rent', 'Median_Rent', 'Pct_Change', 'Units']
        
        # Clean numeric columns (remove letters and ** symbols)
        numeric_cols = ['Vacancy_Rate', 'Avg_Rent', 'Median_Rent', 'Pct_Change', 'Units']
        for col in numeric_cols:
            df[col] = df[col].astype(str).str.replace(r'[a-zA-Z\*\+]', '', regex=True)
            df[col] = pd.to_numeric(df[col], errors='coerce')
        
        df['Year'] = year
        return df
        
    except Exception as e:
        print(f"Error loading {file_path}: {e}")
        return None

# Load all rental data with fixed function
all_rental_data = []
for file in rental_files:
    df = load_rental_data_fixed(file)
    if df is not None:
        all_rental_data.append(df)
        print(f"✅ Loaded {os.path.basename(file)}")

rental_combined = pd.concat(all_rental_data, ignore_index=True)
rental_combined = rental_combined.sort_values(['Year', 'Zone'])

print("\n🏠 Rental data summary:")
print(f"Years: {sorted(rental_combined['Year'].unique())}")
print(f"Zones: {sorted(rental_combined['Zone'].unique())}")

# Show sample data
print("\nSample rental data:")
print(rental_combined[['Zone', 'Year', 'Avg_Rent', 'Vacancy_Rate']].head(10))
```

```python
def load_rental_data_simple(file_path):
    """Load rental data and keep only the good columns"""
    
    # Get the year from filename
    filename = os.path.basename(file_path)
    year = 2000 + int(filename.split()[1])
    
    try:
        # Load the file
        df = pd.read_csv(file_path, encoding='latin-1', skiprows=2)
        
        # Keep only columns that have real names (not "Unnamed")
        good_columns = [col for col in df.columns if 'Unnamed' not in str(col)]
        df = df[good_columns]
        
        # Remove rows that aren't real zones
        df = df[~df.iloc[:, 0].astype(str).str.contains('Notes|CMA|Source|definitions|Calgary', na=False)]
        df = df.dropna(how='all')
        
        # Give the columns nice names
        df.columns = ['Zone', 'Vacancy_Rate', 'Availability_Rate', 'Avg_Rent', 'Median_Rent', 'Pct_Change', 'Units']
        
        # Clean up the numbers (remove letters and symbols)
        for col in ['Vacancy_Rate', 'Avg_Rent', 'Median_Rent', 'Pct_Change', 'Units']:
            df[col] = df[col].astype(str).str.replace(r'[a-zA-Z\*\+]', '', regex=True)
            df[col] = pd.to_numeric(df[col], errors='coerce')
        
        df['Year'] = year
        return df
        
    except Exception as e:
        print(f"Trouble with {file_path}: {e}")
        return None

# Try loading all the rental files now
all_rental_data = []
for file in rental_files:
    df = load_rental_data_simple(file)
    if df is not None:
        all_rental_data.append(df)
        print(f"✅ Got {os.path.basename(file)}")

if all_rental_data:
    rental_combined = pd.concat(all_rental_data, ignore_index=True)
    print(f"\n🏠 We have rental data for {len(all_rental_data)} years!")
    print(rental_combined[['Zone', 'Year', 'Avg_Rent']].head())
```

```python
def load_rental_data_final(file_path):
    """Load rental data understanding the paired column structure"""
    
    # Get year from filename
    filename = os.path.basename(file_path)
    year = 2000 + int(filename.split()[1])
    
    try:
        # Load all columns first
        df = pd.read_csv(file_path, encoding='cp1252', skiprows=2)
        
        # The structure is: Zone, Vacancy%, letter, Availability%, empty, AvgRent$, letter, MedianRent$, letter, %Change, letter, Units, empty
        # Let's pick out just the data columns we want (every other column basically)
        
        # Remove rows with metadata
        df = df[~df.iloc[:, 0].astype(str).str.contains('Notes|CMA|Source|definitions|Calgary', na=False)]
        df = df.dropna(how='all')
        
        # Extract the actual data columns (skip the letter code columns)
        clean_data = {
            'Zone': df.iloc[:, 0],                    # Column 0: Zone names
            'Vacancy_Rate': df.iloc[:, 1],            # Column 1: Vacancy rate
            'Avg_Rent': df.iloc[:, 5],                # Column 5: Average rent  
            'Median_Rent': df.iloc[:, 7],             # Column 7: Median rent
            'Pct_Change': df.iloc[:, 9],              # Column 9: Percent change
            'Units': df.iloc[:, 11]                   # Column 11: Units
        }
        
        df_clean = pd.DataFrame(clean_data)
        
        # Clean up the numeric columns (remove quotes, commas, ++ symbols)
        for col in ['Vacancy_Rate', 'Avg_Rent', 'Median_Rent', 'Pct_Change', 'Units']:
            df_clean[col] = df_clean[col].astype(str).str.replace(r'["\+,]', '', regex=True)
            df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
        
        df_clean['Year'] = year
        return df_clean
        
    except Exception as e:
        print(f"Trouble with {file_path}: {e}")
        return None

# Test with our rental files
all_rental_data = []
for file in rental_files:
    df = load_rental_data_final(file)
    if df is not None:
        all_rental_data.append(df)
        print(f"✅ Got {os.path.basename(file)}")

if all_rental_data:
    rental_combined = pd.concat(all_rental_data, ignore_index=True)
    rental_combined = rental_combined.sort_values(['Year', 'Zone'])
    
    print(f"\n🏠 We have rental data for {len(all_rental_data)} years!")
    print("\nSample data:")
    print(rental_combined[['Zone', 'Year', 'Avg_Rent', 'Vacancy_Rate']].head(10))
```

```python
# Clean up the rental data - remove metadata rows
rental_clean = rental_combined.copy()

# Remove rows where Zone contains metadata symbols or explanations
rental_clean = rental_clean[~rental_clean['Zone'].astype(str).str.contains('\*\*|^\+\+|Data suppressed|Change in rent', na=False)]

# Remove any rows with NaN zone names
rental_clean = rental_clean.dropna(subset=['Zone'])

# Show the clean data
print("🏠 Clean rental data:")
print(rental_clean[['Zone', 'Year', 'Avg_Rent', 'Vacancy_Rate']].head(15))

print(f"\nWe have data for these zones: {sorted(rental_clean['Zone'].unique())}")
print(f"Years: {sorted(rental_clean['Year'].unique())}")
```

```python
# Create a comparison between housing construction and rental trends
print("🔍 Let's compare construction activity to rental market changes!")

# First, let's see which zones had the most housing construction
zone_construction = combined_housing.groupby('Zone').agg({
    'All': 'mean',  # Average monthly construction
    'Apartment': 'mean'
}).round(0)

zone_construction = zone_construction.sort_values('All', ascending=False)
print("\n🏗️ Average monthly construction by zone:")
print(zone_construction.head(10))

# Now let's look at rent changes by zone
rent_changes = rental_clean.pivot_table(
    index='Zone', 
    columns='Year', 
    values='Avg_Rent'
).fillna(method='ffill', axis=1)

# Calculate rent change from 2021 to 2024
rent_changes['Rent_Change_21_24'] = ((rent_changes[2024] - rent_changes[2021]) / rent_changes[2021] * 100).round(1)
rent_changes = rent_changes.sort_values('Rent_Change_21_24')

print("\n💰 Rent changes 2021-2024 by zone:")
print(rent_changes[['Rent_Change_21_24']].head(10))

# Combine construction and rent data
comparison = zone_construction.merge(
    rent_changes[['Rent_Change_21_24']], 
    left_index=True, 
    right_index=True, 
    how='inner'
)

print("\n🎯 THE BIG COMPARISON - High Construction vs Rent Changes:")
print(comparison.sort_values('All', ascending=False))
```

```python
# Create a scatter plot showing construction vs rent changes
fig, ax = plt.subplots(figsize=(12, 8))

# Create the scatter plot
scatter = ax.scatter(comparison['All'], comparison['Rent_Change_21_24'], 
                    s=200, alpha=0.7, c='steelblue', edgecolors='navy')

# Add zone labels to each point
for zone, row in comparison.iterrows():
    ax.annotate(zone, (row['All'], row['Rent_Change_21_24']), 
                xytext=(5, 5), textcoords='offset points', 
                fontsize=10, ha='left')

# Add a trend line
from scipy import stats
slope, intercept, r_value, p_value, std_err = stats.linregress(comparison['All'], comparison['Rent_Change_21_24'])
line = slope * comparison['All'] + intercept
ax.plot(comparison['All'], line, 'r--', alpha=0.8, linewidth=2, 
        label=f'Trend line (R² = {r_value**2:.3f})')

# Styling
ax.set_xlabel('Average Monthly Housing Construction (All Types)', fontsize=12)
ax.set_ylabel('Rent Increase 2021-2024 (%)', fontsize=12)
ax.set_title('Does More Housing Construction = Lower Rent Increases?\nCalgary Zones 2021-2024', 
             fontsize=14, fontweight='bold', pad=20)

ax.grid(True, alpha=0.3)
ax.legend()

# Add interpretation text
if slope < 0:
    interpretation = "✅ YIMBY Theory Confirmed: More construction → Lower rent increases"
    color = 'green'
else:
    interpretation = "❌ YIMBY Theory Challenged: More construction → Higher rent increases"  
    color = 'red'

ax.text(0.02, 0.98, interpretation, transform=ax.transAxes, 
        fontsize=12, fontweight='bold', color=color,
        bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8),
        verticalalignment='top')

plt.tight_layout()
plt.show()

print(f"📊 Statistical relationship:")
print(f"  Correlation coefficient: {r_value:.3f}")
print(f"  R-squared: {r_value**2:.3f}")
print(f"  P-value: {p_value:.3f}")
print(f"  Interpretation: {'Strong' if abs(r_value) > 0.5 else 'Moderate' if abs(r_value) > 0.3 else 'Weak'} {'negative' if r_value < 0 else 'positive'} relationship")
```

```python
# Create before/after rental comparison maps
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(24, 12))
fig.patch.set_facecolor('white')

# Calculate early vs late period rent averages
early_rent = rental_clean[rental_clean['Year'].isin([2021, 2022])].groupby('Zone')['Avg_Rent'].mean().to_dict()
late_rent = rental_clean[rental_clean['Year'].isin([2023, 2024])].groupby('Zone')['Avg_Rent'].mean().to_dict()

# Load map for both subplots
calgary_map = mpimg.imread(img_path)

def create_rent_map(ax, rent_data, title, period_color):
    # Display map background
    ax.imshow(calgary_map, extent=[0, 10, 0, 8], alpha=0.85)
    
    # Get rent range for color coding
    max_rent = max(rent_data.values()) if rent_data else 1500
    min_rent = min(rent_data.values()) if rent_data else 1000
    
    for zone in zone_positions:
        if zone in rent_data:
            x, y = zone_positions[zone]
            rent_value = rent_data[zone]
            
            # Bubble size based on rent level (higher rent = larger bubble)
            base_size = 200
            size_multiplier = ((rent_value - min_rent) / (max_rent - min_rent)) * 600 + base_size
            
            # Color coding: red for high rent, green for low rent
            rent_normalized = (rent_value - min_rent) / (max_rent - min_rent)
            if rent_normalized >= 0.8:
                color = "#C0392B"  # High rent - red
            elif rent_normalized >= 0.6:
                color = "#E74C3C"  # Medium-high rent
            elif rent_normalized >= 0.4:
                color = "#F39C12"  # Medium rent - orange
            elif rent_normalized >= 0.2:
                color = "#F7DC6F"  # Medium-low rent - yellow
            else:
                color = "#27AE60"  # Low rent - green
            
            # Draw bubble
            ax.scatter(x, y, s=size_multiplier, c=color, alpha=0.8, 
                      edgecolors='white', linewidth=2.5, zorder=10)
            
            # Label with rent amount
            ax.annotate(f"{zone}\n${rent_value:.0f}", 
                       (x, y), xytext=(0, -35), textcoords='offset points',
                       fontsize=9, ha='center', va='top', 
                       bbox=dict(boxstyle="round,pad=0.3", facecolor='white', 
                                alpha=0.95, edgecolor='#34495E', linewidth=1))
    
    # Styling
    ax.set_title(title, fontsize=16, fontweight='bold', pad=20, color='#2C3E50')
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 8)
    ax.axis('off')

# Create EARLY PERIOD map (2021-2022)
create_rent_map(ax1, early_rent, 
               "Early Period Rent Levels\n(2021-2022 Average)", "#3498DB")

# Create LATE PERIOD map (2023-2024)  
create_rent_map(ax2, late_rent, 
               "Late Period Rent Levels\n(2023-2024 Average)", "#E67E22")

# Calculate overall rent change
early_avg = sum(early_rent.values()) / len(early_rent)
late_avg = sum(late_rent.values()) / len(late_rent)
rent_pct_change = ((late_avg - early_avg) / early_avg) * 100

# Add overall comparison
fig.text(0.5, 0.02, f"Overall Rent Impact: ${late_avg:.0f} avg rent (2023-24) vs ${early_avg:.0f} avg rent (2021-22) = +{rent_pct_change:.1f}% increase", 
         ha='center', fontsize=14, fontweight='bold', color='#2C3E50')

# Create rent legend
rent_legend_elements = []
rent_ranges = [
    ("Very High ($1500+)", "#C0392B", 800),
    ("High ($1400-1499)", "#E74C3C", 650),
    ("Medium ($1300-1399)", "#F39C12", 500),
    ("Low ($1200-1299)", "#F7DC6F", 350),
    ("Very Low (<$1200)", "#27AE60", 200)
]

for label, color, size in rent_ranges:
    rent_legend_elements.append(plt.scatter([], [], s=size, c=color, alpha=0.8, 
                                          edgecolors='white', linewidth=2))

fig.legend(rent_legend_elements, [label.split(' (')[0] for label, _, _ in rent_ranges],
          scatterpoints=1, loc='center', bbox_to_anchor=(0.5, 0.08), ncol=5, 
          frameon=True, fontsize=11, title='Average Monthly Rent Level')

plt.suptitle('Calgary Rental Market Evolution\nComparing Average Rent Levels: Early Period vs Late Period', 
             fontsize=18, fontweight='bold', y=0.95)

plt.tight_layout()
plt.subplots_adjust(bottom=0.15)
plt.show()

print("💰 RENTAL MARKET COMPARISON:")
print(f"Early period avg rent (2021-22): ${early_avg:.0f}")
print(f"Late period avg rent (2023-24): ${late_avg:.0f}")
print(f"Overall rent change: +{rent_pct_change:.1f}%")

print("\nZone-by-zone rent changes:")
for zone in zone_positions:
    if zone in early_rent and zone in late_rent:
        change = late_rent[zone] - early_rent[zone]
        pct_change = (change / early_rent[zone]) * 100
        print(f"{zone}: ${early_rent[zone]:.0f} → ${late_rent[zone]:.0f} ({pct_change:+.1f}%)")
```

```python
# Create a comprehensive Housing Affordability Index
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 10))
fig.patch.set_facecolor('white')

# Calculate comprehensive affordability scores for each zone
affordability_scores = {}

for zone in zone_positions:
    if (zone in comparison.index and 
        zone in rent_increases and 
        zone in rental_clean['Zone'].values):
        
        # Get metrics
        construction_rate = comparison.loc[zone, 'All']
        rent_increase_rate = rent_increases[zone]
        
        # Get vacancy rate (higher = more affordable)
        zone_vacancy = rental_clean[rental_clean['Zone'] == zone]['Vacancy_Rate'].mean()
        
        # Create affordability score (0-100, higher = more affordable)
        # Factors: High construction (+), Low rent growth (+), High vacancy (+)
        
        # Normalize each metric (0-1 scale)
        construction_norm = min(construction_rate / 500, 1.0)  # Cap at 500 units
        rent_growth_norm = max(0, 1 - (rent_increase_rate / 60))  # Inverse: lower growth = better
        vacancy_norm = min(zone_vacancy / 10, 1.0)  # Cap at 10% vacancy
        
        # Weighted affordability score
        affordability_score = (
            construction_norm * 40 +      # 40% weight on new supply
            rent_growth_norm * 40 +       # 40% weight on rent stability  
            vacancy_norm * 20             # 20% weight on availability
        ) * 100
        
        affordability_scores[zone] = {
            'score': affordability_score,
            'construction': construction_rate,
            'rent_growth': rent_increase_rate,
            'vacancy': zone_vacancy
        }

# Sort by affordability score
sorted_zones = sorted(affordability_scores.items(), key=lambda x: x[1]['score'], reverse=True)

# Create the affordability ranking chart
zones = [item[0] for item in sorted_zones]
scores = [item[1]['score'] for item in sorted_zones]

# Color code: Green for high affordability, Red for low
colors = []
for score in scores:
    if score >= 70:
        colors.append('#27AE60')      # High affordability
    elif score >= 50:
        colors.append('#F39C12')      # Medium affordability
    else:
        colors.append('#E74C3C')      # Low affordability

bars = ax1.barh(range(len(zones)), scores, color=colors, alpha=0.8, edgecolor='white', linewidth=2)

# Customize the ranking chart
ax1.set_yticks(range(len(zones)))
ax1.set_yticklabels(zones, fontsize=11)
ax1.set_xlabel('Housing Affordability Index Score (0-100)', fontsize=12, fontweight='bold')
ax1.set_title('Calgary Housing Affordability Rankings\nCombined Index: Construction + Rent Growth + Availability', 
              fontsize=14, fontweight='bold', pad=20)

# Add score labels on bars
for i, (bar, score) in enumerate(zip(bars, scores)):
    ax1.text(score + 1, i, f'{score:.0f}', va='center', fontweight='bold', fontsize=10)

# Add ranking numbers
for i, zone in enumerate(zones):
    rank = i + 1
    medal = "A" if rank == 1 else "B" if rank == 2 else "C" if rank == 3 else f"#{rank}"
    ax1.text(-5, i, medal, va='center', ha='right', fontsize=12, fontweight='bold')

ax1.grid(axis='x', alpha=0.3)
ax1.set_xlim(0, 100)

# Create the detailed breakdown chart  
metrics = ['Construction Rate', 'Rent Stability', 'Availability']
top_3_zones = zones[:3]

x = np.arange(len(metrics))
width = 0.25

for i, zone in enumerate(top_3_zones):
    data = affordability_scores[zone]
    
    # Convert to component scores
    construction_component = (min(data['construction'] / 500, 1.0)) * 40
    rent_component = (max(0, 1 - (data['rent_growth'] / 60))) * 40
    vacancy_component = (min(data['vacancy'] / 10, 1.0)) * 20
    
    values = [construction_component, rent_component, vacancy_component]
    
    bars = ax2.bar(x + i*width, values, width, label=zone, alpha=0.8)

ax2.set_xlabel('Affordability Components', fontsize=12, fontweight='bold')
ax2.set_ylabel('Component Score (0-40/20)', fontsize=12, fontweight='bold')
ax2.set_title('Top 3 Zones: Affordability Breakdown\nWhat Makes Them Successful?', 
              fontsize=14, fontweight='bold', pad=20)
ax2.set_xticks(x + width)
ax2.set_xticklabels(metrics)
ax2.legend(title='Top Performing Zones', title_fontsize=11, fontsize=10)
ax2.grid(axis='y', alpha=0.3)

# Add overall insights
insights_text = f"""🏆 AFFORDABILITY CHAMPIONS:
1st: {zones[0]} (Score: {scores[0]:.0f})
2nd: {zones[1]} (Score: {scores[1]:.0f})  
3rd: {zones[2]} (Score: {scores[2]:.0f})

💡 KEY INSIGHT:
Zones with high construction AND low rent 
growth create the most affordable housing"""

fig.text(0.02, 0.02, insights_text, fontsize=11, 
         bbox=dict(boxstyle="round,pad=0.5", facecolor='#E8F6F3', alpha=0.9),
         verticalalignment='bottom')

plt.suptitle('Calgary Housing Affordability Index Analysis\nWhich Zones Offer the Best Value for Residents?', 
             fontsize=16, fontweight='bold', y=0.95)

plt.tight_layout()
plt.show()

# Print detailed results
print(" HOUSING AFFORDABILITY INDEX RESULTS:")
print("="*50)
for i, (zone, data) in enumerate(sorted_zones):
    rank = i + 1
    medal = "A" if rank == 1 else "B" if rank == 2 else "C" if rank == 3 else f"#{rank}"
    print(f"{medal} {zone}: {data['score']:.1f} points")
    print(f"   Construction: {data['construction']:.0f} units/mo")
    print(f"   Rent growth: {data['rent_growth']:.1f}%")
    print(f"   Vacancy rate: {data['vacancy']:.1f}%")
    print()

print(" This affordability index shows Calgary's policy created real value for residents!")
print("The top-ranked zones combine strong housing supply with reasonable rent growth.")
```

```python
# Create small multiples: Each zone's individual housing construction story
fig, axes = plt.subplots(2, 5, figsize=(20, 10))
fig.patch.set_facecolor('white')
axes = axes.flatten()

# Get monthly data for each zone
zones = sorted(combined_housing['Zone'].unique())

# Academic color palette
line_color = '#2E4057'
pre_color = '#D6EAF8'
post_color = '#A9DFBF'
accent_color = '#E74C3C'

for i, zone in enumerate(zones):
    ax = axes[i]
    
    # Get zone data
    zone_data = combined_housing[combined_housing['Zone'] == zone].copy()
    zone_data = zone_data.sort_values('YearMonth')
    
    # Split pre/post rezoning
    pre_data = zone_data[zone_data['YearMonth'] < '2024-08']
    post_data = zone_data[zone_data['YearMonth'] >= '2024-08']
    
    # Plot the line with different colors for pre/post
    if len(pre_data) > 0:
        ax.plot(range(len(pre_data)), pre_data['All'], 
                color=line_color, linewidth=2, alpha=0.8)
        ax.fill_between(range(len(pre_data)), pre_data['All'], 
                       alpha=0.3, color=pre_color)
    
    if len(post_data) > 0:
        start_idx = len(pre_data)
        ax.plot(range(start_idx, start_idx + len(post_data)), post_data['All'], 
                color=accent_color, linewidth=3, alpha=0.9)
        ax.fill_between(range(start_idx, start_idx + len(post_data)), post_data['All'], 
                       alpha=0.4, color=post_color)
    
    # Add rezoning line
    if len(pre_data) > 0:
        ax.axvline(x=len(pre_data), color=accent_color, linestyle='--', 
                  linewidth=2, alpha=0.7)
    
    # Calculate impact
    pre_avg = pre_data['All'].mean() if len(pre_data) > 0 else 0
    post_avg = post_data['All'].mean() if len(post_data) > 0 else 0
    change_pct = ((post_avg - pre_avg) / pre_avg * 100) if pre_avg > 0 else 0
    
    # Style each subplot
    ax.set_title(f"{zone}\n{change_pct:+.1f}% change", 
                fontsize=11, fontweight='bold', pad=10)
    ax.set_ylim(0, max(zone_data['All']) * 1.1 if len(zone_data) > 0 else 100)
    
    # Clean minimal axes
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_color('#CCCCCC')
    ax.spines['bottom'].set_color('#CCCCCC')
    ax.grid(True, alpha=0.2)
    ax.tick_params(axis='x', which='both', bottom=False, top=False, labelbottom=False)
    ax.tick_params(axis='y', labelsize=8, colors='#666666')
    
    # Add subtle annotation for high performers
    if change_pct > 50:
        ax.text(0.95, 0.95, '🚀', transform=ax.transAxes, 
               fontsize=12, ha='right', va='top')

# Overall title and subtitle
fig.suptitle('Individual Zone Stories: Calgary Housing Construction Evolution', 
             fontsize=18, fontweight='bold', y=0.95)
fig.text(0.5, 0.91, 'Each chart shows monthly housing starts by zone, with policy implementation marked in red', 
         ha='center', fontsize=12, style='italic', color='#666666')

# Create custom legend
legend_elements = [
    plt.Line2D([0], [0], color=line_color, linewidth=2, label='Pre-Rezoning'),
    plt.Line2D([0], [0], color=accent_color, linewidth=3, label='Post-Rezoning'),
    plt.Line2D([0], [0], color=accent_color, linewidth=2, linestyle='--', label='Policy Implementation')
]

fig.legend(handles=legend_elements, loc='lower center', ncol=3, 
          bbox_to_anchor=(0.5, 0.02), fontsize=11, frameon=False)

# Add insight box
insight_text = """Key Insights:
- Every zone shows policy response
- Northwest leads with largest increase  
- Even low-construction zones show uplift
- Policy created city-wide housing momentum"""

fig.text(0.02, 0.25, insight_text, fontsize=10, 
         bbox=dict(boxstyle="round,pad=0.5", facecolor='#F8F9FA', 
                   edgecolor='#BDC3C7', alpha=0.95),
         verticalalignment='center')

plt.tight_layout()
plt.subplots_adjust(top=0.85, bottom=0.15)
plt.show()

print("📊 Small multiples visualization: Individual zone housing stories!")
print("This shows how Calgary's rezoning policy created momentum across ALL zones,")
print("not just the high-performing ones. Each zone has its own trajectory!")
```
